```python
!pip install -U google-genai
from google import genai

client = genai.Client(api_key="AQ. ")

MODEL = "models/gemini-3.5-flash"
```

```python
teacher_info = {
    "BSCS-6A": "Miss. Naheed Azeem",
    "BSCS-6B": "Dr. Muhammad Sarim",
    "BSCS-6C": "Dr. Syed Akhtar Raza",
    "BSCS-6D": "Dr. Uzma Afzal",
    "BSCS-6E": "Miss. Sana"
}


def university_agent(user_input):

    prompt = f"""
You are a helpful chatbot for Federal Urdu University.

Your main tasks are:

1. TEACHER INFORMATION
- If the user asks about a specific section, check the Teacher Directory.
- Give only:
  Teacher Name
  Section
- Do not give extra teacher details.
- If the section is not available in the directory, say:
  "Sorry, teacher information for this section is not available."

2. GENERAL UNIVERSITY INFORMATION
- For general questions about Federal Urdu University, answer clearly and simply.
- You may use your general knowledge to answer.
- If you do not know the answer, clearly say that the information is not available.
- Do not make up university information.

Teacher_info:
{teacher_info}

User Question:
{user_input}

Answer:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("Federal Urdu University Chatbot Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye!")
        break

    try:
        reply = university_agent(user)
        print("\nChatbot:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
