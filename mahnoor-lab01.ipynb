```python
print("Welcome to Pakistan AI Travel Agent")
name = input("Enter your name: ")
budget = int(input("Enter your travel budget (in PKR): "))

print(f"\nHello {name}, finding a trip within your budget of Rs.{budget}...\n")

if budget < 10000:
    print("Your budget is too low for a trip. Try saving a bit more!")

elif budget >= 10000 and budget < 30000:
    print("Suggested Destination: Murree")
    print("Why: Nearby hill station, low travel cost, great weather.")
    print("Estimated cost: Rs.10,000 - Rs.30,000")

elif budget >= 30000 and budget < 60000:
    print("Suggested Destination: Naran & Kaghan Valley")
    print("Why: Scenic valleys, rivers, moderate travel cost.")
    print("Estimated cost: Rs.30,000 - Rs.60,000")

elif budget >= 60000 and budget < 100000:
    print("Suggested Destination: Hunza Valley")
    print("Why: Mountains, lakes, and one of the most beautiful places in Pakistan.")
    print("Estimated cost: Rs.60,000 - Rs.100,000")

elif budget >= 100000 and budget < 200000:
    print("Suggested Destination: Skardu & Deosai Plains")
    print("Why: Premium experience, high mountains, lakes, and unique landscapes.")
    print("Estimated cost: Rs.100,000 - Rs.200,000")

else:
    print("Suggested Destination: Full Northern Pakistan Tour (Hunza + Skardu + Naran)")
    print("Why: With this budget you can cover multiple regions comfortably.")
    print("Estimated cost: Rs.200,000+")

print("\nThank you for using Pakistan AI Travel Agent. Have a safe trip!")
```
